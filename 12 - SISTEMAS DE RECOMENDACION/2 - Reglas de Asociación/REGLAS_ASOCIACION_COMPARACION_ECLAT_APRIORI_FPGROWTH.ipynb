```python
# Conteo de ítems por cantidad de transacciones
items_total = super_bin.sum().sort_values(ascending=False)

# Crear DataFrame con estadísticas
df_table = pd.DataFrame({
    'items': items_total.index,
    'transactions': items_total.values
}).sort_values("transactions", ascending=False)

# Mostrar los 20 más populares con gradiente azul
df_table.head(20).style.background_gradient(cmap='Blues')
```

```python

# Agregar columna raíz para el TreeMap
df_table["all"] = "Tree Map"

# Crear TreeMap interactivo (top 50 ítems)
fig = px.treemap(
    df_table.head(50),
    path=['all', "items"],
    values='transactions',
    color='transactions',
    hover_data=['items'],
    color_continuous_scale='Blues'
)

# Mostrar TreeMap
fig.show()
```

```python
# DEMORA MÁS DE 55 MINUTOS con max_combination=7
# the item shoud appear at least at 5% of transactions
min_support = 2/100
# start from transactions containing at least 2 items
min_combination = 2
# up to maximum items per transaction
#max_combination = max(items_per_transaction)      # = 7

# pyEclat entrega las reglas calculando sólo el Soporte. Es más rápido pero da pocas métricas...

# Ajustar el modelo ECLAT (solo calcula Soporte)
rule_indices, rule_supports = eclat.fit(
    min_support=min_support,           # soporte mínimo (2%)
    min_combination=min_combination,   # mínimo 2 ítems por regla
    max_combination=2,                 # máximo 2 ítems por regla (acá es rápido)
    separator=', ',                    # separador en los ítems
    verbose=True                       # muestra barra de progreso
)
```

COMPARACIÓN CON APRIORI

```python

# Definir soporte mínimo = 2%
min_support = 2/100  

# Obtener itemsets frecuentes a partir del DataFrame binario (super_bin)
apriori_items = apriori(
    super_bin,            # DataFrame one-hot (binario: 0/1 por ítem)
    use_colnames=True,    # que los nombres de los ítems aparezcan en los resultados
    min_support=min_support,
    max_len=2             # hasta combinaciones de 2 ítems
)

# Mostrar cuántos itemsets fueron encontrados
print(apriori_items.shape)

# Ver los itemsets frecuentes con su soporte
apriori_items
```

```python
from mlxtend.frequent_patterns import association_rules

# Generar reglas de asociación usando lift como métrica
reglas = association_rules(apriori_items, metric="lift", min_threshold=1.0)

# Mostrar dimensiones
print(reglas.shape)
print("TODAS LAS ASOCIACIONES ENCONTRADAS")

# Ordenar por lift (descendente) y resetear índice
reglas = reglas.sort_values(['lift'], ascending=False).reset_index(drop=True)
reglas
```

COMPARACIÓN CON FP-GROWTH


```python
# https://towardsdatascience.com/fp-growth-frequent-pattern-generation-in-data-mining-with-python-implementation-244e561ab1c3
# https://rasbt.github.io/mlxtend/user_guide/frequent_patterns/fpgrowth/

# fpgrowth(df, min_support=0.5, use_colnames=False, max_len=None, verbose=0)
# Get frequent itemsets from a one-hot DataFrame
max_combination=2
fp_items = fpgrowth(super_bin, min_support=min_support, use_colnames=True, max_len=max_combination)
fp_items
```

```python
reglas_fp = association_rules(fp_items, metric = 'leverage', min_threshold=0.01, num_itemsets=2)
print(reglas_fp.shape)
print('TODAS LAS ASOCIACIONES ENCONTRADAS')
reglas=reglas_fp.sort_values(['lift'], ascending=False).reset_index(drop=True)
reglas
```
