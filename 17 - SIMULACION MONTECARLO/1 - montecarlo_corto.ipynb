```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Created on Fri Sep  3 18:40:01 2021
@author: anahi
"""

import random  # Importamos el módulo 'random' para generar números aleatorios

# ---------------------------------------------------------
# Función que genera un punto (x, y) con coordenadas aleatorias entre 0 y 1
# ---------------------------------------------------------
def generar_punto():
    x = random.random()  # Genera un número aleatorio entre 0 y 1
    y = random.random()  # Genera otro número aleatorio entre 0 y 1
    #print(x, y)          # (opcional) se podría imprimir el punto generado
    return x, y          # Devuelve el par (x, y)

# ---------------------------------------------------------
# Función que estima el valor de π usando el método de Monte Carlo
# ---------------------------------------------------------
def estimar_pi(n):
    m = 0                       # Contador de puntos dentro del círculo
    for i in range(n):           # Repite el experimento 'n' veces
        x, y = generar_punto()   # Genera un punto aleatorio
        # Ecuación del círculo: x² + y² <= 1 (radio = 1)
        if x**2 + y**2 <= 1:     # Si el punto cae dentro del cuarto de círculo
            m += 1               # Incrementa el contador
    # Fórmula de estimación:
    # (puntos dentro del círculo / total de puntos) ≈ área del cuarto de círculo (π/4)
    pi_est = 4 * m / n           # Se multiplica por 4 porque solo simulamos un cuadrante
    return pi_est                # Devuelve la estimación de π

# ---------------------------------------------------------
# Probamos la función con diferentes cantidades de puntos
# ---------------------------------------------------------
casi_pi = [10, 100, 1000, 10000, 100000, 1000000, 10000000]

# Para cada cantidad de puntos, se imprime una estimación de π
for c in casi_pi:
    print(f'Pi está cerca de: {estimar_pi(c)}')
```

```python
import random  # Importamos el módulo 'random' para generar números aleatorios

def calculopi(n):
    # -----------------------------------------------------
    # Calcula una estimación de π usando el método Monte Carlo
    # -----------------------------------------------------

    r = 1.0                 # Radio del círculo (unidad)
    a, b = (0.0, 0.0)       # Centro del círculo en el origen (0,0)

    # Límites del cuadrado circunscrito al círculo de radio 1
    x_neg, x_pos = a - r, a + r  # desde -1 hasta 1 en eje X
    y_neg, y_pos = b - r, b + r  # desde -1 hasta 1 en eje Y

    count = 0.0             # Contador de puntos dentro del círculo

    # -----------------------------------------------------
    # Generamos 'n' puntos aleatorios dentro del cuadrado
    # -----------------------------------------------------
    for i in range(0, n):

        # Genera un valor aleatorio de X e Y entre los límites del cuadrado
        x = random.uniform(x_neg, x_pos)  # distribución uniforme entre -1 y 1
        y = random.uniform(y_neg, y_pos)

        # Verifica si el punto (x, y) cae dentro del círculo de radio 1
        if x*x + y*y <= 1.0:
            count += 1  # Si está dentro, incrementa el contador

    # -----------------------------------------------------
    # Estimación de π:
    # (puntos dentro del círculo / total puntos) ≈ área círculo / área cuadrado
    # área del círculo = πr², área del cuadrado = (2r)² = 4r² → proporción ≈ π/4
    # -----------------------------------------------------
    result = (count / float(n)) * 4

    # Imprime el resultado estimado
    print("result is", result)

# -----------------------------------------------------
# Pruebas con diferentes cantidades de puntos
# -----------------------------------------------------
calculopi(10)        # Muy poca precisión
calculopi(1000)      # Un poco mejor
calculopi(1000000)   # Bastante buena aproximación
calculopi(100000000) # Muy precisa (pero puede tardar)
```

```python
import random

def zero2onetox2(n):
    # calcular integral definida para f(x) = x**2 con x entre 0 y 1, su valor es 1/3
    #n = 1000000
    x_min,x_max = 0.0,1.0
    y_min,y_max = 0.0,1.0
    count = 0

    for i in range(0,n):
        x = random.uniform(x_min,x_max)
        y = random.uniform(y_min,y_max)

        if x*x >= y:
            count += 1

    result = count / float(n)
    print('n=', n)
    print("result is ",result)
    return result

print(f'Diferencia= {(1/3-zero2onetox2(10))*100}%')
print(f'Diferencia= {(1/3-zero2onetox2(1000))*100}%')
print(f'Diferencia= {(1/3-zero2onetox2(1000000))*100}%')
```
