```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Created on Fri Sep  3 18:40:01 2021

@author: anahi
"""

import random

def generar_punto():
    x = random.random()    # producen nros seudo aleatorios entre 0 y 1
    y = random.random()
    #print(x,y)

    return x,y

def estimar_pi(n):
    m = 0
    for i in range(n):
        x, y = generar_punto()
        if x**2 + y**2 <= 1:
            m += 1
    pi_est = 4 * m/n    # se multiplica *4  xq sólo estamos mirando un cuarto de círculo

    return pi_est

casi_pi = [10, 100, 1000, 10000, 100000, 1000000, 10000000]
for c in casi_pi:
    print(f'Pi está cerca de: {estimar_pi(c)}')

```

```python
import random

def calculopi(n):
    #calcular pi usando Montecarlo
    #n = 1000000
    r = 1.0
    a, b = (0.0, 0.0)
    x_neg, x_pos = a - r, a + r
    y_neg, y_pos = b - r, b + r

    count = 0.0
    for i in range(0, n):

        x = random.uniform(x_neg, x_pos)    # distribución uniforme entre 0 y 1
        y = random.uniform(y_neg, y_pos)

        if x*x + y*y <= 1.0:
            count += 1

    result = (count / float(n)) * 4
    print ("result is ",result)

calculopi(10)
calculopi(1000)
calculopi(1000000)
calculopi(100000000)
```

```python
import random

def zero2onetox2(n):
    # calcular integral definida para f(x) = x**2 con x entre 0 y 1, su valor es 1/3
    #n = 1000000
    x_min,x_max = 0.0,1.0
    y_min,y_max = 0.0,1.0
    count = 0

    for i in range(0,n):
        x = random.uniform(x_min,x_max)
        y = random.uniform(y_min,y_max)

        if x*x >= y:
            count += 1

    result = count / float(n)
    print('n=', n)
    print("result is ",result)
    return result

print(f'Diferencia= {(1/3-zero2onetox2(10))*100}%')
print(f'Diferencia= {(1/3-zero2onetox2(1000))*100}%')
print(f'Diferencia= {(1/3-zero2onetox2(1000000))*100}%')
```
