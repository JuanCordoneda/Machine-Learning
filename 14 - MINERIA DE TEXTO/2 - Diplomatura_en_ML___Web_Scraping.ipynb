1) Importar Librerias necearias (este ejemplo podemos trabajarlo con BeautifulSoup, sin necesidad de Selenium)



```python
import requests
import urllib.request
import re
from bs4 import BeautifulSoup
import pandas as pd
import requests
```

2) Primero obtener la url sección de Economía del portal (no hace falta hacerlo con scraping, vale buscarla manualmente), y aplicar un bucle que guarde todos los links de las notas

```python
#Primero obtenemos los links de la sección
page = requests.get("https://www.lapoliticaonline.com.ar/seccion/economia/")

soup = BeautifulSoup(page.content, 'html.parser', from_encoding='utf8')

lista=[]

for a in soup.find_all('a', href=True):
   lista.append(a['href'])
```

```python
lista
```

3) Filrar los links de las notas (excluir links a la home, a otras secciones, etc.).

*Asegurarse de que no queden links duplicados (esto se puede evitar con la función set())*

```python
lista= [x for x in lista if '/nota' in x and 'https' not in x]
lista_filtrada = list(set(lista_filtrada))
```

```python
lista_filtrada
```

4) Recorrer cada link y scrapear el texto de las notas. Almacenar en una lista el texto y la url correspondiente a cada nota.

```python
texto=[]
nota=[]

for i in lista_filtrada[0:4]: #Cambiar el número para sumar más notas
  page = requests.get("https://www.lapoliticaonline.com.ar/"+i)
  soup = BeautifulSoup(page.content, 'html.parser')

  for a in soup.findAll(class_= "body vsmcontent"):
    texto.append(a.text)
    nota.append(i)
```

5) Construir un Dataframe que tenga en cada fila la url y el texto

```python
df= pd.DataFrame(zip(nota,texto))
```
