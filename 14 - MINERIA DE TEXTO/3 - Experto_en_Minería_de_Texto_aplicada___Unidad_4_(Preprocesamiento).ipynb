# Contar frecuencia

Una vez que obtuvimos nuestra lista de tokens, lo siguiente que vamos a hacer es generar el vocabulario del texto. Vamos a usar la lista "tokens_reg" que tiene todos los tokens menos los símbolos.

Si nuestra lista de tokens contiene cada aparición de un string compatible con una palabra (o la unidad que busquemos),
un vocabulario va a contener:
una sola aparición de cada palabra (set) en nuestro texto, o
un conteo de la frecuencia de cada palabras (defaultdict) en nuestro texto.

```python
#@title
set(tokens_reg)
from collections import defaultdict

frecuencia = defaultdict(int)

for token in tokens_reg:
    frecuencia[token] += 1

print(frecuencia)
```

# Stemming

🌱 1. ¿Qué es un stem?

Un stem es la raíz de una palabra, lo que queda después de quitar sufijos, terminaciones o desinencias.
👉 La idea es que distintas palabras relacionadas (flexiones o derivados) se reduzcan a la misma forma base.

Ejemplo en español:

libro, libreria, librera, libreta → todas tienen como raíz libr.

🔧 2. Cómo funciona tu código

En tu caso usás un regex (expresión regular) como “stemmer casero”:

stemmer = re.compile(r'[aeiou]\b|eria\b|er[oa]\b|eta\b')


Eso significa:

[aeiou]\b → elimina vocales al final de palabra.

eria\b → elimina el sufijo -eria.

er[oa]\b → elimina -ero o -era.

eta\b → elimina -eta.

```python
# Importamos 're' para expresiones regulares y defaultdict para contar fácilmente
import re
from collections import defaultdict

# Definimos un "stemmer" casero usando regex
# Elimina vocales al final de palabra (\b = fin de palabra)
# y sufijos comunes como 'eria', 'ero/a', 'eta'
stemmer = re.compile(r'[aeiou]\b|eria\b|er[oa]\b|eta\b')

# Lista de palabras de ejemplo
texto = ['libro', 'libreria', 'librera', 'libreta']

# Diccionario para contar formas léxicas (palabras originales)
formas_lexicas = defaultdict(int)

# Diccionario para contar las raíces (stems)
stems = defaultdict(int)

# Recorremos cada palabra del texto
for i in texto:
    # Contamos la forma original
    formas_lexicas[i] += 1

    # Aplicamos el regex para obtener la raíz/stem
    raiz = stemmer.sub('', i)

    # Contamos la raíz
    stems[raiz] += 1

# Mostramos las palabras originales con sus conteos
print("Formas léxicas:", dict(formas_lexicas))

# Mostramos las raíces obtenidas con sus conteos
print("Stems:", dict(stems))
```

# Preprocesamiento con NLTK

# Pipeline de Preprocesamiento

En este código se construye una clase con todos los pasos vistos anteriormente.


```python
import nltk
import re
import string
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer

# Descargamos recursos necesarios
nltk.download('stopwords')
nltk.download('punkt')

# Stopwords y stemmer en español
default_stopwords = stopwords.words('spanish')
default_stemmer = SnowballStemmer("spanish")

def clean_text(text):
    """ Preprocesa texto en español: lowercase, tokeniza, quita stopwords, stemming, elimina caracteres especiales """

    # Tokenización en palabras
    def tokenize_text(text):
        return [w for s in sent_tokenize(text) for w in word_tokenize(s)]

    # Eliminar caracteres especiales (mantener letras y números)
    def remove_special_characters(text, characters=string.punctuation.replace('-', '')):
        tokens = tokenize_text(text)
        pattern = re.compile('[{}]'.format(re.escape(characters)))
        return ' '.join(filter(None, [pattern.sub('', t) for t in tokens]))

    # Aplicar stemming
    def stem_text(text, stemmer=default_stemmer):
        tokens = tokenize_text(text)
        return ' '.join([stemmer.stem(t) for t in tokens])

    # Eliminar stopwords
    def remove_stopwords(text, stop_words=default_stopwords):
        tokens = [w for w in tokenize_text(text) if w not in stop_words]
        return ' '.join(tokens)

    # ==== Pipeline de limpieza ====
    text = text.strip()          # quitar espacios extremos
    text = text.lower()          # minúsculas
    text = remove_special_characters(text)  
    text = remove_stopwords(text)
    text = stem_text(text)

    return text
```

SpaCy parsea y etiqueta los tokens.


---


**Text:** Texto de la palabra original.
Lemma: Forma base de la palabra

**POS:** POS tag simple.

**Tag:** POS tag detallado.

**Dep:** Dependencia sintáctica.

**Shape:** Forma de la palabra - puntuación, dígitos, mayúsculas.

**is alpha:** ¿El token es alfabético?

**is stop:** ¿El token es una stopword?

```python
import pandas as pd
from spacy.tokens import Doc
tokens_info = {'text':[],
              'lemma':[],
              'pos':[],
              'tag':[],
              'dep':[],
              'shape':[],
              'is_alpha':[],
              'is_stop':[]}


for token in doc:
    tokens_info['text'].append(token.text)
    tokens_info['lemma'].append(token.lemma_)
    tokens_info['pos'].append(token.pos_)
    tokens_info['tag'].append(token.tag_)
    tokens_info['dep'].append(token.dep_)
    tokens_info['shape'].append(token.shape_)
    tokens_info['is_alpha'].append(token.is_alpha)
    tokens_info['is_stop'].append(token.is_stop)

df = pd.DataFrame(tokens_info)
df.iloc[50:65,:]
```

# Matcheo mediante reglas

Rules-base matcher es un componente de SpaCy que posibilita generar reglas con las que contrastar el texto y extraer los fragmentos que se adecúen a ellas. Lo interesante de este componente es que permite combinar expresiones regulares con las anotaciones lingüísticas de los modelos pre-entrenados.

```python
import re

#doc = nlp(novela)
sentences_good = list()

i = 1
for sent in doc.sents:
    sentences_good.append(sent.text.strip())
    i += 1
```

```python
from spacy.matcher import Matcher
from spacy import displacy
matcher = Matcher(nlp.vocab)

#patternSV = [{"ORTH": "Aureliano", "DEP":'nsubj'},{"POS": "VERB", "DEP": "ROOT"},]

patternSV = [{"ORTH": "Aureliano", "DEP":'nsubj'},{"TEXT":{"REGEX":"\w+"},"OP":"*"},{"POS": "VERB", "DEP": "ROOT"},]
#patternSV = [{"ORTH": "tierra"},{"IS_PUNCT": True}]

matcher.add("patternSV", [patternSV])

matched_sents = list()

for sent in sentences_good[1:900]: #sentences_good[-5:-1]:
    s = nlp(sent)
    matches = matcher(s)
    match_ents = list()
    if len(matches) > 0:
        for match in matches:
            match_id, start, end = match
            span = s[start:end]
            info_match = {
                "start": span.start_char,
                "end": span.end_char,
                "label": "MATCH"
            }
            match_ents.append(info_match)
        matched_sents.append({"text": sent, "ents": match_ents})


colors = {"MATCH": "linear-gradient(90deg, #b7deed, #49a5bf)"}
options = {"ents": ["MATCH"], "colors": colors}
displacy.render(matched_sents, style="ent", options=options, jupyter=True, manual=True)
```

## NER

*Un* NER (Named Entity Recognition) es un sistema que no solo nos permite encontrar o reconocer entidades sino, además, saber de qué tipo son.

Ahora bien, ¿qué es una entidad? ¿Cuántos tipos de entidades hay o puede haber? ¿Para qué puede servirnos reconocer entidades?

El NER de SpaCy nos permite reconocer los siguientes atributos de una entidad:

**Text:** texto original de la entidad

**Start:** índice de inicio de la entidad en el Doc

**End:** índice de finalización de la entidad en el Doc

**LabeL:** etiqueta de la entidad (tipo de entidad)

```python
print('{0:50} {1:10} {2:2} {3}'.format('text','start','end','label'),end='\n\n')

for sent in sentences_good[59:72]:
    s = nlp(sent)
    for ent in s.ents:
        print('{0:50} {1:5} {2:8} {3}'.format(ent.text.strip(), ent.start_char, ent.end_char, ent.label_))
```
