```python
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import numpy as np
df = pd.read_csv('EDMHits.csv')

df.shape
```

Renombramiento de columnas al español

```python
# Renombrar columnas a español
df = df.rename(columns={
    "Track": "Cancion",
    "Artist": "Artista",
    "Album": "Album",
    "Year": "Año",
    "Duration": "Duracion_ms",
    "Time_Signature": "Compas",
    "Danceability": "Bailabilidad",
    "Energy": "Energia",
    "Key": "Tonalidad",
    "Loudness": "Volumen",
    "Mode": "Modo",
    "Speechiness": "Habla",
    "Acousticness": "Acustica",
    "Instrumentalness": "Instrumentalidad",
    "Liveness": "Presencia_en_vivo",
    "Valence": "Valencia",
    "Tempo": "Tempo",
    "Popularity": "Popularidad"
})

# Verificar el resultado
df.head()
```

```python
import numpy as np

# Asegurar tipo de dato correcto
df["Año"] = df["Año"].astype(int)

# Reemplazar ceros en Tonalidad por NaN y luego por la moda
df["Tonalidad"].replace(0, np.nan, inplace=True)
df["Tonalidad"].fillna(df["Tonalidad"].mode()[0], inplace=True)

# Crear columna binaria de éxito (1 = popularidad >= 60)
df["Exito"] = (df["Popularidad"] >= 60).astype(int)

# Ver distribución de clases
print(df["Exito"].value_counts(normalize=True))

# Copia para visualización
df_viz = df.copy()

# Eliminar columnas no predictivas
df_modelo = df.drop(columns=["Cancion", "Artista", "Album"])

# Eliminar filas con valores faltantes
df_modelo.dropna(inplace=True)

print("✅ Dataset listo. Filas:", df_modelo.shape[0])
```

```python
# Separar variables independientes (X) y dependiente (y)
X = df_modelo.drop(columns=["Popularidad", "Exito"])
y = df_modelo["Exito"]

# Mostrar las primeras filas
print("Tamaño del dataset:", X.shape)
X.head()
```
