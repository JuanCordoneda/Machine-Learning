```python
import numpy as np      # Librería para manejo numérico y creación de arrays
import matplotlib.pyplot as plt   # Librería para graficar funciones
import math             # Librería matemática (para usar coseno y pi)

# 💡 DEFINIMOS LA FUNCIÓN A MAXIMIZAR
def fx(x):
    # Esta función combina una parte cuadrática (1-x)^2 y una parte oscilante (coseno)
    # El término -(...) invierte la forma de la función (para buscar máximos en lugar de mínimos)
    # El +2 al final solo eleva toda la función para que quede positiva (desplazamiento vertical)
    return -(0.1 + (1 - x)**2 - 0.1 * math.cos(6 * math.pi * (1 - x))) + 2

# Creamos una lista vacía para guardar los valores de y = f(x)
y_axis = []

# Generamos los valores de x desde 0 hasta 2, con incrementos de 0.02
# np.arange(inicio, fin, paso) → crea una secuencia de números
x_axis = np.arange(0, 2, 0.02)

# 🔁 Recorremos cada valor de x y calculamos su correspondiente y
for num in x_axis:
    # Calculamos fx(num) y lo agregamos a la lista de valores de y
    y_axis.append(fx(num))

# 📈 Graficamos la función
plt.plot(x_axis, y_axis)    # Dibuja la curva (x vs fx(x))
plt.title("Gráfico de la función fx(x)")  # Título del gráfico
plt.xlabel("x")             # Etiqueta del eje X
plt.ylabel("f(x)")          # Etiqueta del eje Y
plt.grid(True)              # Muestra una cuadrícula de referencia
plt.show()                  # Muestra el gráfico en pantalla
```

En pocas palabras:

👉 El “fitness” mide qué tan bueno es un individuo dentro de la población.

En un algoritmo genético:

Cada individuo representa una posible solución (por ejemplo, un número o combinación de parámetros).

La función de fitness evalúa esa solución, asignándole un valor de éxito o calidad.

Cuanto mayor es el valor de fitness, mejor es el individuo.

📈 Por ejemplo:
Si querés maximizar una función f(x), entonces el fitness puede ser directamente f(x).
Los individuos con valores de f(x) más altos tienen más probabilidad de ser seleccionados para reproducirse y generar nuevas soluciones.

En resumen:

🔹 El fitness es la medida de qué tan buena es una solución en comparación con las demás.

```python
# Definimos la función de evaluación o "fitness"
def fitness(poblacion):
    '''Calcula la aptitud (fitness) de cada individuo en la población'''
    fitness = []  # lista para guardar los valores de fitness
    
    # Recorre cada individuo para calcular su valor de éxito
    for individuo in poblacion:
        x = listToDecimal(individuo)  # convierte el individuo en número decimal
        y = fx(x)                     # evalúa la función objetivo con ese valor
        fitness += [y]                # guarda el resultado
    
    # Convierte la lista en un vector numpy para operaciones matemáticas eficientes
    fitness = np.array(fitness)
    
    # Normaliza los valores de fitness dividiendo por la suma total → resultados entre 0 y 1
    fitness = fitness / fitness.sum()
    
    return fitness  # devuelve los valores de aptitud normalizados

# Calculamos el fitness de la población actual
ftns = fitness(poblacion)
print(type(ftns))  # muestra el tipo (numpy.ndarray)
ftns.sum()         # debería dar 1.0 (por la normalización)
```

💡 Qué hace esta parte del código

Esta función simula la “reproducción” en un algoritmo genético, o sea, cómo se crean nuevos individuos (nuevas soluciones) combinando partes de dos anteriores.

🧠 Paso a paso:

Tenés una población inicial, que son varios individuos (por ejemplo, listas de números que representan posibles soluciones).
👉 Ejemplo:

[0, 7, 2, 1, 5]
[1, 4, 9, 8, 3]
[0, 5, 6, 7, 2]


Cada individuo tiene un “fitness”, que indica qué tan bueno es (qué tan bien resuelve el problema).
Los mejores individuos tienen más probabilidad de ser elegidos como padres.

Elegís dos padres al azar, pero con más chances para los que tienen mejor fitness:

parents = np.random.choice(size_poblacion, 2, p=fitness)


— Esto devuelve dos índices de padres elegidos probabilísticamente.

Se elige un punto de cruce (“cross_point”), por ejemplo el índice 3.

Se cruzan los padres:

Hijo 1 = primera parte del padre A + segunda parte del padre B

Hijo 2 = primera parte del padre B + segunda parte del padre A

Ejemplo visual:

Padre A: [0, 7, 2 | 1, 5]
Padre B: [1, 4, 9 | 8, 3]

Hijos:
[0, 7, 2, 8, 3]
[1, 4, 9, 1, 5]


Así se generan todos los hijos que forman la nueva generación (next_gen), que reemplazará o se combinará con la anterior.

🧩 En resumen:

La función reproduction() mezcla el ADN (los números) de dos individuos con mejor fitness para crear nuevas soluciones.
Es como una versión digital de la reproducción biológica: cruce + azar = evolución.

```python
#  DEFINO LA OPERACIÓN DE CRIANZA X CROSSPOINT
# https://numpy.org/doc/stable/reference/random/generated/numpy.random.choice.html

def reproduction(poblacion, ind_size, fitness):
    size_poblacion = len(poblacion)
    #hijos
    children = []
    for i in range(size_poblacion//2):
        parents = np.random.choice(size_poblacion, 2, p=fitness)
        #print(parents)  # queremos ver cuáles índices se toman
        cross_point = np.random.randint(ind_size)
        children += [poblacion[parents[0]][:cross_point] + poblacion[parents[1]][cross_point:]]
        children += [poblacion[parents[1]][:cross_point] + poblacion[parents[0]][cross_point:]]

    return children

next_gen = reproduction(poblacion, ind_size, ftns)
print(next_gen[:3])
```

```python
#  VEMOS A LA SGTE GENERACIÓN
poblacion = next_gen
graficar(poblacion)
```

🧬 Qué hace la función mutate()

La función introduce pequeñas modificaciones aleatorias (mutaciones) en los individuos de la población.
Es una parte esencial de los algoritmos genéticos, porque evita que todos los individuos terminen siendo iguales y se pierda la “diversidad genética”.

```python
#   DEFINIMOS LA OPERACIÓN DE MUTACIÓN CON UNA PROBABILIDAD MUY BAJA

def mutate(individuals, prob, pool):
    for i in range(len(individuals)):
        mutate_individual=individuals[i]
        if np.random.random() < prob:
            mutation = np.random.choice(pool[0])
            mutate_individual = [mutation] + mutate_individual[1:]

        for j in range(1,len(mutate_individual)):
            if np.random.random() < prob:
                mutation = np.random.choice(pool[1])
                mutate_individual = mutate_individual[0:j] + [mutation] + mutate_individual[j+1:]
        individuals[i] = mutate_individual

    return individuals
```

```python
#  AHORA JUNTAMOS TODO Y HACEMOS EVOLUCIONAR A LA POBLACIÓN DURANTE UNA
#  CANTIDAD DE GENERACIONES

def evolucionar(poblacion, generaciones, prob_mut):
    ''' Hace evolucionar una población durante una cantidad de generaciones'''

    for _ in range(generaciones):
        fit = fitness(poblacion)
        # REPRODUCCIÓN
        children = reproduction(poblacion, ind_size, fit)
        # MUTACIONES
        poblacion = mutate(children,prob_mut,genetic_pool)

    return poblacion
```
