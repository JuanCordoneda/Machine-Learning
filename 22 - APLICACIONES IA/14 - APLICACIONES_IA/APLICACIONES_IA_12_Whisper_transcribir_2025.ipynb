### Audio a texto



**4. Creamos una función de transcripción**

```python
def transcribe_audio(cliente, ruta_archivo):
    # Abre el archivo de audio en modo lectura binaria ('rb' significa read binary)
    # Esto es necesario para poder enviar el archivo en formato correcto a la API
    archivo_audio = open(ruta_archivo, "rb")

    # Llama a la API del cliente Whisper para transcribir el archivo de audio
    # `cliente.audio.transcriptions.create` es el método que envía el archivo a la API de Whisper
    # Se le pasa el modelo de Whisper a utilizar ("whisper-1") y el archivo de audio
    transcripcion = cliente.audio.transcriptions.create(
        model="whisper-1",  # Especifica el modelo de Whisper a usar
        file=archivo_audio  # El archivo de audio abierto se pasa aquí
    )

    # Devuelve el resultado de la transcripción generado por la API de Whisper
    return transcripcion
```

### Transcribiendo a otros idiomas

**6. Ahora creamos función de traducción de audio a texto**

```python
def traduce_audio_a_ingles(cliente, ruta_archivo):
    # Define una función llamada 'traduce_audio_a_ingles' que toma dos argumentos:
    # 'cliente' (el cliente que interactúa con el servicio de traducción de audio) y
    # 'ruta_archivo' (la ruta donde se encuentra el archivo de audio que se quiere traducir).

    archivo_audio = open(ruta_archivo, "rb")
    # Abre el archivo de audio ubicado en 'ruta_archivo' en modo de lectura binaria ("rb").
    # Esto es necesario para trabajar con archivos de audio ya que suelen almacenarse en formato binario.
    # 'archivo_audio' ahora contiene una referencia al archivo abierto.

    traduccion = cliente.audio.translations.create(
        model="whisper-1",
        file=archivo_audio) #, language='french'

    # Utiliza el cliente proporcionado (posiblemente una instancia de la API de OpenAI u otro servicio)
    # para crear una traducción del archivo de audio.
    # La llamada 'cliente.audio.translations.create' usa el modelo "whisper-1" para hacer la traducción.
    # 'file=archivo_audio' pasa el archivo de audio abierto como entrada para el proceso de traducción.
    # El resultado de la traducción (probablemente texto traducido del audio) se almacena en la variable 'traduccion'.

    return traduccion
    # Finalmente, la función devuelve el resultado de la traducción.
```

```python
#Ejecuto la función
traduccion_1 = traduce_audio_a_ingles(client, "prueba_whisper.mp3")
traduccion_1
```

```python
from txtai.pipeline import Translation

def traduce_audio_a_otro_idioma(cliente, ruta_archivo, idioma):
    # Define una función llamada 'traduce_audio_a_ingles' que toma dos argumentos:
    # 'cliente' (el cliente que interactúa con el servicio de traducción de audio) y
    # 'ruta_archivo' (la ruta donde se encuentra el archivo de audio que se quiere traducir).

    archivo_audio = open(ruta_archivo, "rb")
    # Abre el archivo de audio ubicado en 'ruta_archivo' en modo de lectura binaria ("rb").
    # Esto es necesario para trabajar con archivos de audio ya que suelen almacenarse en formato binario.
    # 'archivo_audio' ahora contiene una referencia al archivo abierto.

    transcripcion = cliente.audio.transcriptions.create(
        model="whisper-1",
        file=archivo_audio)

    # Utiliza el cliente proporcionado (posiblemente una instancia de la API de OpenAI u otro servicio)
    # para crear una traducción del archivo de audio.
    # La llamada 'cliente.audio.translations.create' usa el modelo "whisper-1" para hacer la traducción.
    # 'file=archivo_audio' pasa el archivo de audio abierto como entrada para el proceso de traducción.
    # El resultado de la traducción (probablemente texto traducido del audio) se almacena en la variable 'traduccion'.

    # Crear modelo de traducción de txtai para pasar el texto a otro idioma
    translate = Translation()
    translation = translate(transcripcion.text, idioma) # 'idioma' debe ser un string

    return translation
    # Finalmente, la función devuelve el resultado de la traducción.
```

```python
traduccion_2 = traduce_audio_a_otro_idioma(client, "prueba_whisper.mp3", 'it')
traduccion_2
```

Creamos una transcripción de texto a audio con Whisper

[Fuente](https://github.com/openai/openai-python/blob/main/examples/audio.py)

```python
speech_file_path='audio.mp3'
def main() -> None:
    # Create text-to-speech audio file
    with client.audio.speech.with_streaming_response.create(
        model="tts-1",
        voice="alloy",
        input="the quick brown fox jumped over the lazy dogs",
    ) as response:
        response.stream_to_file(speech_file_path)

main()
```
