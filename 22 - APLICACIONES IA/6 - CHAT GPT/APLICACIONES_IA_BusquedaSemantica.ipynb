Definimos la función **BuscarComentarios(Opiniones,Consulta,n)** que busca las similitudes (coseno) entre una **consulta** y las **Opiniones**, y entrega los **n** meuores resultados. Para esto, la consulta se convierte a su correspondiente embedding:

```python
# ===============================================================
# Función para calcular la similitud entre textos usando embeddings
# ===============================================================

# Calcula la similitud coseno entre dos vectores usando NumPy
def cosine_similarity(a, b):
    # Fórmula: cos(θ) = (a · b) / (||a|| * ||b||)
    return dot(a, b) / (norm(a) * norm(b))


# ---------------------------------------------------------------
# BuscarComentarios:
# Dada una consulta y una lista de opiniones con embeddings,
# devuelve los n comentarios más similares (por similitud coseno).
# ---------------------------------------------------------------
def BuscarComentarios(Opiniones, consulta, n=3):
    # 1️⃣ Inicializa el cliente de OpenAI con la API key
    client = OpenAI(api_key=openai_api_key)

    # 2️⃣ Obtiene el embedding de la consulta usando el modelo de embeddings
    response = client.embeddings.create(
        model=MODELO_EMBEDDING,
        input=consulta
    )

    # Extrae el vector de embedding resultante
    EmbeddingConsulta = response.data[0].embedding

    # 3️⃣ Calcula la similitud coseno entre la consulta y cada opinión del DataFrame
    # Se agrega una nueva columna "similitud" con el valor correspondiente
    Opiniones["similitud"] = Opiniones.embedding.apply(
        lambda x: cosine_similarity(np.array(x), np.array(EmbeddingConsulta))
    )

    # 4️⃣ Ordena las opiniones de mayor a menor similitud y devuelve las n más parecidas
    results = Opiniones.sort_values("similitud", ascending=False).head(n)
    return results
```

```python
# ------------------------------------------------------------
# Selección del modelo de embeddings y carga de datos
# ------------------------------------------------------------

# 1️⃣ Elegimos el modelo de embeddings de OpenAI
# "text-embedding-ada-002" es un modelo muy eficiente para representar texto en vectores numéricos.
# Captura el significado semántico de frases, palabras o documentos.
MODELO_EMBEDDING = "text-embedding-ada-002"

# 2️⃣ Cargamos el archivo CSV con los comentarios (en inglés)
# Se asume que la primera columna (índice 0) contiene el texto de cada comentario.
Opiniones = pd.read_csv('/content/sentiments-ingles.csv', header=None)

# 3️⃣ Inicializamos el cliente de OpenAI con nuestra clave de API.
# Esto permite hacer solicitudes al modelo de embeddings.
client = OpenAI(api_key=openai_api_key)
```

```python
# Define a helper function to get embeddings for a text
def get_embedding_for_text(text, model):
    response = client.embeddings.create(
        model=model,
        input=text
    )
    return response.data[0].embedding

Opiniones["embedding"] = Opiniones[0].apply(lambda x: get_embedding_for_text(x, MODELO_EMBEDDING))
```
