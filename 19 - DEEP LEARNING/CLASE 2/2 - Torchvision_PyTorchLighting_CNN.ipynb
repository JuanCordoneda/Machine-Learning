### <span style="color:blue">Crear clase para los datos</span>

Los DataModule en PyTorch Lightning permiten separar la preparación y organización de los datos del proceso de entrenamiento de un modelo. Esto significa que podemos manipular nuestros datos de manera independiente al modelo de entrenamiento, lo que facilita la reutilización y el intercambio de datos sin necesidad de modificar el código del modelo.

Con `DataModule` podemos organiza la canalización de datos en una clase compartible y reutilizable. Un módulo de datos encapsula los cinco pasos involucrados en el procesamiento de datos en PyTorch:

* Descargar/tokenizar/procesar.
* Limpiar y (tal vez) guardar en el disco.
* Cargar dentro del conjunto de datos.
* Aplicar transformaciones (rotar, tokenizar, etc.).
* Envolver dentro de un DataLoader.

Obtenga más información sobre los módulos de datos [aquí](https://pytorch-lightning.readthedocs.io/en/latest/data/datamodule.html).

Construimos una clase derivada de `LightningDataModule` específica para nuestro conjunto de datos.

```python
# Definición de la clase CIFAR10DataModule que hereda de pl.LightningDataModule
class CIFAR10DataModule(pl.LightningDataModule):

    # Hiperparámetros y transformaciones para la canalización de datos
    def __init__(self, batch_size, data_dir: str = './', num_workers=4):
    # la elección 4 es muy común porque muchas CPUs modernas tienen al menos 4 núcleos,
    #lo que permite un buen equilibrio entre rendimiento y uso de recursos sin saturar el sistema.
        super().__init__()  # Llama al inicializador de la clase base LightningDataModule
        self.data_dir = data_dir  # Directorio donde se encuentran los datos
        self.batch_size = batch_size  # Tamaño de lote para los dataloaders
        self.num_workers = num_workers  # Número de procesadores para cargar los datos en paralelo
        # Transformaciones que se aplicarán a los datos
        self.transform = transforms.Compose([
            transforms.ToTensor(),  # Convierte los datos en tensores PyTorch
            transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
            # Normaliza los valores de los píxeles, con medias y desviaciones estándar para cada canal (rojo, verde, azul)
        ]) #los valores del conjunto estan entre 1 y 0, y con esto estaria entre -1 y 1 lo cual ayuda a mejorar el modelo
        self.dims = (3, 32, 32)  # Dimensiones de la imagen (3 canales: RGB, 32x32 píxeles)
        self.num_classes = 10  # Número de clases en el conjunto de datos CIFAR-10

    # Método para descargar los datos si no están presentes en el directorio especificado
    def prepare_data(self):
        # Descarga los datos de entrenamiento (train=True) en el directorio self.data_dir
        CIFAR10(self.data_dir, train=True, download=True)
        # Descarga los datos de prueba (train=False) en el directorio self.data_dir
        CIFAR10(self.data_dir, train=False, download=True)

    # Método para preparar los conjuntos de datos en tensores PyTorch para cada división de los datos
    def setup(self, stage=None): # se coloca none porque no importa en que estado este el modelo para hacer los procesos de training
        # Asigna datos a los datasets de entrenamiento y validación si la etapa es 'fit' o None (ambas)
        if stage == 'fit' or stage is None:
            # Carga los datos de entrenamiento y aplica las transformaciones
            cifar_full = CIFAR10(self.data_dir, train=True, transform=self.transform)
            # Divide los datos de entrenamiento en 45000 ejemplos para entrenamiento y 5000 para validación
            self.cifar_train, self.cifar_val = random_split(cifar_full, [45000, 5000])
        # Asigna dataset de prueba si la etapa es 'test' o None (ambas)
        if stage == 'test' or stage is None:
            # Carga los datos de prueba y aplica las transformaciones
            self.cifar_test = CIFAR10(self.data_dir, train=False, transform=self.transform)

    # Método para crear el DataLoader de entrenamiento
    def train_dataloader(self):
        # Crea y devuelve un DataLoader para el conjunto de entrenamiento, con barajado de datos (shuffle=True)
        return DataLoader(self.cifar_train, batch_size=self.batch_size, shuffle=True)

    # Método para crear el DataLoader de validación
    def val_dataloader(self):
        # Crea y devuelve un DataLoader para el conjunto de validación, sin barajado de datos
        return DataLoader(self.cifar_val, batch_size=self.batch_size)

    # Método para crear el DataLoader de prueba
    def test_dataloader(self):
        # Crea y devuelve un DataLoader para el conjunto de prueba, sin barajado de datos
        return DataLoader(self.cifar_test, batch_size=self.batch_size)
```

Vamos a inicializar nuestra canalización de datos, de una vez. Dicha inicialización se podría haber hecho luego de crear las demás clases que necesitamos, pero hacerlo de esta manera nos sirve para el proceso pedagógico y para probar que la clase que acabamos de crear, en términos generales, está funcionando bien.

Mas adelante, el objeto `dm` que hemos creado podría pasar directamente al `Trainer`; pero como necesitamos algunas muestras para un callback personalizado que implementaremos, llamaremos manualmente a los métodos `prepare_data` y `setup`.

```python
# Inicializar nuestra canalización de datos
dm = CIFAR10DataModule(batch_size=32)  # Crea una instancia de CIFAR10DataModule con tamaño de lote 32

# Para acceder al x_dataloader necesitamos llamar a prepare_data y setup.
dm.prepare_data()  # Descarga los datos del conjunto CIFAR-10 si no están ya presentes en el directorio especificado
dm.setup()  # Configura los conjuntos de datos de entrenamiento, validación y prueba
```

Ahora produciremos una muestra de lo que requerirá nuestro callback personalizado para registrar las predicciones de las imágenes:

```python
val_samples = next(iter(dm.val_dataloader()))
# Obtiene el primer lote de datos del dataloader de validación.
# dm.val_dataloader() devuelve un DataLoader para el conjunto de validación.
# iter(dm.val_dataloader()) crea un iterador sobre los lotes del dataloader.
# next(...) obtiene el primer lote del iterador.

val_imgs, val_labels = val_samples[0], val_samples[1]
# Desempaqueta el lote en imágenes y etiquetas.
# val_samples[0] contiene las imágenes del lote.
# val_samples[1] contiene las etiquetas correspondientes a las imágenes.

val_imgs.shape, val_labels.shape
# Muestra las formas (dimensiones) de las imágenes y etiquetas del lote.
# val_imgs.shape devuelve la forma del tensor de imágenes.
# val_labels.shape devuelve la forma del tensor de etiquetas.
```

### <span style="color:blue">Crear clase para callback personalizado</span>

Un callback es un componente reutilizable que se puede aplicar en múltiples proyectos para realizar tareas específicas durante el entrenamiento de un modelo. PyTorch Lightning incluye varios callbacks integrados que se usan con frecuencia, y puedes aprender más sobre ellos [aquí](https://pytorch-lightning.readthedocs.io/en/latest/extensions/callbacks.html).


En el contexto de la clasificación de imágenes, visualizar las predicciones del modelo en un subconjunto de imágenes puede ser extremadamente útil. Este tipo de visualización ayuda a depurar y entender el comportamiento del modelo en las primeras etapas del desarrollo. Para lograr esto, vamos a implementar un callback personalizado que registre las predicciones del modelo en el panel de control de Weights & Biases (W&B, wandb).

La clase ImagePredictionLogger hereda de la clase Callback de PyTorch Lightning. En su constructor, se pasa val_samples, una tupla que contiene imágenes de validación y sus etiquetas correspondientes. Además, se especifica num_samples, que determina el número de imágenes que se registrarán en el panel de control de W&B. Este callback se activará al final de cada época de validación, moviendo las imágenes y etiquetas al dispositivo adecuado, obteniendo las predicciones del modelo y registrando las imágenes junto con sus predicciones y etiquetas correctas en W&B.

```python
# Definición de la clase ImagePredictionLogger que hereda de Callback.
# Este callback se utiliza para registrar predicciones del modelo al final de cada época de validación.
class ImagePredictionLogger(Callback):
    def __init__(self, val_samples, num_samples=32):
        super().__init__()  # Inicializa la clase base Callback.
        self.num_samples = num_samples  # Número de muestras de validación a registrar.
        self.val_imgs, self.val_labels = val_samples  # Almacena las imágenes de validación y etiquetas.

    def on_validation_epoch_end(self, trainer, pl_module):
        val_imgs = self.val_imgs.to(device=pl_module.device)  # Mueve las imágenes de validación al dispositivo del módulo de entrenamiento.
        val_labels = self.val_labels.to(device=pl_module.device)  # Mueve las etiquetas de validación al dispositivo del módulo de entrenamiento.
        logits = pl_module(val_imgs)  # Obtiene las predicciones del modelo para las imágenes de validación.
        preds = torch.argmax(logits, -1)  # Obtiene las predicciones finales del modelo tomando el índice con mayor valor en la última dimensión (-1).
        # -1 se refiere a la última dimensión del tensor, que corresponde a las diferentes clases en la predicción.
        #Usar -1 permite que el código sea independiente del número exacto de dimensiones del tensor.
        # Registra las imágenes y sus predicciones en el logger.
        trainer.logger.experiment.log({
            "examples": [wandb.Image(x, caption=f"Pred:{pred}, Label:{y}")  # Crea un objeto de imagen con la predicción y la etiqueta correcta.
                         for x, pred, y in zip(val_imgs[:self.num_samples],  # Itera sobre las primeras num_samples imágenes.
                                               preds[:self.num_samples],  # Predicciones correspondientes a las primeras num_samples imágenes.
                                               val_labels[:self.num_samples])]  # Etiquetas correspondientes a las primeras num_samples imágenes.
        })
```

El método `on_validation_epoch_end` se llama cuando termina la validación en cada época. Toma dos argumentos: `trainer` y `pl_module`, que se pasan automáticamente por el `Trainer`.
Al usar `trainer.logger.experimental`, podemos utilizar todas las características disponibles de Weights & Biases.

### <span style="color:blue">Crear clase para el sistema</span>

Con `LightningModule` definimos un sistema y no un modelo. Aquí, un sistema agrupa todo el código en una sola clase para que sea autónomo. `LightningModule` organiza su código *PyTorch* en 5 secciones:

* Cálculos y elementos del modelo (`__init__`)
* Bucle de entranamiento (`training_step`)
* Bucle de validación (`validation_step`)
* Bucle de prueba (`test_step`)
* Optimizadores (`configure_optimizers`)

Creamos una clase propia asociada al sistema deseado:

La configuración de capas convolucionales con filtros de tamaño 3x3 y pasos de 1 es una práctica común en arquitecturas de CNN exitosas como VGG y ResNet, que han demostrado ser efectivas para la extracción de características en imágenes

```python
class LitModel(pl.LightningModule):
    def __init__(self, input_shape, num_classes, learning_rate=3e-4):
        super().__init__()  # Inicializa la clase base LightningModule.
        self.save_hyperparameters()  # Guarda los hiperparámetros para su uso posterior.
        self.learning_rate = learning_rate  # Tasa de aprendizaje para el optimizador. 3e-4 (0.0003) es un valor comúnmente usado que balancea
        # bien la velocidad de convergencia y la estabilidad del entrenamiento, evitando pasos demasiado grandes que puedan desestabilizar el modelo.
        self.input_shape = input_shape  # Forma de las imágenes de entrada, por ejemplo, (3, 32, 32) para imágenes de CIFAR-10.
        self.num_classes = num_classes  # Número de clases de salida, por ejemplo, 10 para CIFAR-10.

        # Definición de las capas del modelo.
        self.conv1 = nn.Conv2d(3, 32, 3, 1)  # Primera capa convolucional: 3 canales de entrada (RGB), 32 filtros, tamaño del filtro 3x3, paso 1.
        self.conv2 = nn.Conv2d(32, 32, 3, 1)  # Segunda capa convolucional: 32 canales de entrada, 32 filtros, tamaño del filtro 3x3, paso 1.
        self.conv3 = nn.Conv2d(32, 64, 3, 1)  # Tercera capa convolucional: 32 canales de entrada, 64 filtros, tamaño del filtro 3x3, paso 1.
        self.conv4 = nn.Conv2d(64, 64, 3, 1)  # Cuarta capa convolucional: 64 canales de entrada, 64 filtros, tamaño del filtro 3x3, paso 1.
        # MaxPool operación en que reduce las dimensiones espaciales (alto y ancho) de la entrada, manteniendo las características más importantes.
        self.pool1 = nn.MaxPool2d(2)  # Primera capa de Max Pooling: tamaño del kernel 2x2, reduce las dimensiones de las características a la mitad.
        self.pool2 = nn.MaxPool2d(2)  # Segunda capa de Max Pooling: tamaño del kernel 2x2, reduce las dimensiones de las características a la mitad.
        n_sizes = self._get_conv_output(self.input_shape)  # Obtiene el tamaño de salida de las capas convolucionales.
        self.fc1 = nn.Linear(n_sizes, 512)  # Primera capa completamente conectada: entradas del tamaño calculado, 512 neuronas de salida.
        self.fc2 = nn.Linear(512, 128)  # Segunda capa completamente conectada: 512 entradas, 128 neuronas de salida.
        self.fc3 = nn.Linear(128, self.num_classes)  # Tercera capa completamente conectada: 128 entradas, número de clases de salida.
        self.accuracy = torchmetrics.Accuracy(task="multiclass", num_classes=self.num_classes)  # Métrica de precisión para clasificación multiclase.

    # Método para obtener el tamaño de salida de las capas convolucionales.
    def _get_conv_output(self, shape):
        batch_size = 1  # Tamaño de lote ficticio para la prueba.
        input = torch.autograd.Variable(torch.rand(batch_size, *shape))  # Crea un tensor de entrada aleatorio con la forma especificada.
        output_feat = self._forward_features(input)  # Pasa el tensor a través de las capas convolucionales.
        n_size = output_feat.data.view(batch_size, -1).size(1)  # Calcula el tamaño de salida aplanando el tensor.
        return n_size  # Devuelve el tamaño de salida.

    # Método para calcular las características de las capas convolucionales.
    def _forward_features(self, x):
        x = F.relu(self.conv1(x))  # Aplica la primera capa convolucional seguida de una activación ReLU.
        x = self.pool1(F.relu(self.conv2(x)))  # Aplica la segunda capa convolucional seguida de ReLU y Max Pooling.
        x = F.relu(self.conv3(x))  # Aplica la tercera capa convolucional seguida de una activación ReLU.
        x = self.pool2(F.relu(self.conv4(x)))  # Aplica la cuarta capa convolucional seguida de ReLU y Max Pooling.
        return x  # Devuelve el tensor de características.

    # Método forward del modelo.
    def forward(self, x):
        x = self._forward_features(x)  # Propagación hacia adelante a través de las capas convolucionales.
        x = x.view(x.size(0), -1)  # Aplana el tensor de características. el cero será 32 el tamaño del batct y el -1 lo aplana
        # y algo como esto  (32, 4096), donde 4096 = 64 * 8 * 8.
        x = F.relu(self.fc1(x))  # Propagación hacia adelante a través de la primera capa completamente conectada con activación ReLU.
        x = F.relu(self.fc2(x))  # Propagación hacia adelante a través de la segunda capa completamente conectada con activación ReLU.
        x = F.log_softmax(self.fc3(x), dim=1)  # Propagación hacia adelante a través de la tercera capa completamente conectada con activación log-softmax.
        #aca el softmax a función softmax toma un vector de valores  los convierte en probabilidades, es decir, valores entre 0 y 1 que suman 1.
        #
        #El uso de log_softmax en lugar de aplicar softmax seguido de log mejora la estabilidad numérica y evita problemas de desbordamiento.
        return x  # Devuelve las predicciones del modelo.

    # Método de paso de entrenamiento.
    def training_step(self, batch, batch_idx):
        x, y = batch  # Obtiene los datos y etiquetas del lote.
        logits = self(x)  # Predicciones del modelo.
        loss = F.nll_loss(logits, y)  # Cálculo de la función de pérdida de entropía cruzada negativa.
        preds = torch.argmax(logits, dim=1)  # Predicciones de clases tomando el índice con mayor valor en la última dimensión.
        acc = self.accuracy(preds, y)  # Cálculo de la precisión.
        # Registro de métricas de entrenamiento.
        self.log('train_loss', loss, on_step=True, on_epoch=True, logger=True)  # Registra la pérdida de entrenamiento.
        self.log('train_acc', acc, on_step=True, on_epoch=True, logger=True)  # Registra la precisión de entrenamiento.
        return loss  # Devuelve la pérdida para el paso de optimización.

    # Método de paso de validación.
    def validation_step(self, batch, batch_idx):
        x, y = batch  # Obtiene los datos y etiquetas del lote.
        logits = self(x)  # Predicciones del modelo.
        loss = F.nll_loss(logits, y)  # Cálculo de la función de pérdida de entropía cruzada negativa.
        preds = torch.argmax(logits, dim=1)  # Predicciones de clases tomando el índice con mayor valor en la última dimensión.
        acc = self.accuracy(preds, y)  # Cálculo de la precisión.
        # Registro de métricas de validación.
        self.log('val_loss', loss, prog_bar=True)  # Registra la pérdida de validación.
        self.log('val_acc', acc, prog_bar=True)  # Registra la precisión de validación.
        return loss  # Devuelve la pérdida para el paso de validación.

    # Método de paso de prueba.
    def test_step(self, batch, batch_idx):
        x, y = batch  # Obtiene los datos y etiquetas del lote.
        logits = self(x)  # Predicciones del modelo.
        loss = F.nll_loss(logits, y)  # Cálculo de la función de pérdida de entropía cruzada negativa.
        preds = torch.argmax(logits, dim=1)  # Predicciones de clases tomando el índice con mayor valor en la última dimensión.
        acc = self.accuracy(preds, y)  # Cálculo de la precisión.
        # Registro de métricas de prueba.
        self.log('test_loss', loss, prog_bar=True)  # Registra la pérdida de prueba.
        self.log('test_acc', acc, prog_bar=True)  # Registra la precisión de prueba.
        return loss  # Devuelve la pérdida para el paso de prueba.

    # Configuración del optimizador.
    def configure_optimizers(self):
        optimizer = torch.optim.Adam(self.parameters(), lr=self.learning_rate)  # Optimizador Adam con la tasa de aprendizaje especificada.
        return optimizer  # Devuelve el optimizador.
```

Ya hemos inicializado la canalización de los datos en una sección anterior, entonces aquí sólo necesitamos inicializar el modelo y el *logger* para hacer el seguimiento.

```python
# Inicializa el modelo (instancia la clase que acabamos de crear)
model = LitModel(dm.dims, dm.num_classes)
```

```python
# Inicializa *logger* de wandb
wandb_logger = WandbLogger(project='wandb-lightning', job_type='train')
```

### <span style="color:blue">Entrenamiento y evaluación</span>

Ahora que hemos organizado el flujo de datos, la arquitectura del modelo y el bucle de entrenamiento, el `Trainer` de PyTorch Lightning automatiza todo lo demás para nosotros.

El `Trainer` automatiza:

* Iteración de época y lote
* Llamadas a `optimizer.step()`, `backward` y `zero_grad()`
* Llamadas a `.eval()`, habilitando/deshabilitando gradientes
* Guardado y carga de pesos
* Registro de pesos y medidas con Weights & Biases
* Soporte para entrenamiento en múltiples GPU
* Soporte para TPU
* Soporte para entrenamiento en 16 bits

Obtenga más información sobre el `Trainer` [aquí](https://pytorch-lightning.readthedocs.io/en/stable/common/trainer.html).

```python
# Inicializa el Trainer
checkpoint_callback = pl.callbacks.ModelCheckpoint()  # Callback para guardar el mejor modelo basado en la métrica de validación.
early_stop_callback = pl.callbacks.EarlyStopping('val_loss')  # Callback para detener el entrenamiento temprano
#si la métrica de validación ('val_loss') no mejora.
trainer = pl.Trainer(
    max_epochs=50,  # Número máximo de épocas para entrenar el modelo.
    accelerator="gpu",  # Utiliza GPU para acelerar el entrenamiento.
    logger=wandb_logger,  # Logger de Weights & Biases para registrar métricas y visualizaciones.
    callbacks=[early_stop_callback,  # Callback para detener el entrenamiento temprano.
               ImagePredictionLogger(val_samples),  # Callback personalizado para registrar predicciones del modelo en el conjunto de validación.
               checkpoint_callback]  # Callback para guardar el mejor modelo.
)
```

```python
# Entrena el modelo
trainer.fit(model, dm)
```

```python
# Evalua el modelo usando el conjunto de datos de prueba (test)
trainer.test(dataloaders=dm.test_dataloader()) # automáticamente carga los mejores pesos del entrenamiento anterior
```

```python
# Le da finalizar al *logger* de wandb
wandb.finish()
```
