# ¿Qué es el Deep Learning?

El aprendizaje profundo, o "deep learning", es una rama de la inteligencia artificial que utiliza redes neuronales profundas para modelar y resolver tareas complejas. Se destaca por su capacidad para aprender automáticamente a través de grandes conjuntos de datos sin una programación explícita. Este enfoque ha demostrado excelentes resultados en tareas como reconocimiento de imágenes, procesamiento del lenguaje natural y toma de decisiones complejas. Los paréntesis () son parte de la sintaxis de llamada de función en Python.
La diferencia clave con un vector radica en la dimensionalidad. En este caso, tienes tensores bidimensionales (matrices) porque están definidos con dos corchetes [[]]. Cada fila en la matriz representa un vector unidimensional, y la matriz en su conjunto forma un tensor bidimensional.

````python
zeros = torch.zeros(2, 3)         # crea un tensor 2x3 lleno de ceros
print(zeros)                      # muestra el tensor de ceros

ones = torch.ones(2, 3)           # crea un tensor 2x3 lleno de unos
print(ones)                       # muestra el tensor de unos

torch.manual_seed(1729)           # fija la semilla para reproducibilidad en los valores aleatorios
random = torch.rand(2, 3)         # genera un tensor 2x3 con valores aleatorios uniformes en [0, 1)
print(random)                     # muestra el tensor aleatorio generado
````

## Formas de los Tensores

A menudo, para poder realizar operaciones en dos o más tensores, estos tendrán que tener la misma forma, es decir, tener el mismo número de dimensiones y el mismo número de celdas en cada dimensión. Para garantizar eso existen los métodos torch.*_like(): Usando el atributo .shape, podemos verificar que cada uno de estos métodos devuelve un tensor de dimensionalidad y extensión idénticas.

````python
x = torch.empty(2, 2, 3)            # tensor sin inicializar con forma tridimensional 2x2x3
print(x.shape)
print(x)

empty_like_x = torch.empty_like(x)  # tensor vacío con misma forma y tipo que x
print(empty_like_x.shape)
print(empty_like_x)
````

### Tipos de datos de un Tensor

En PyTorch, puedes especificar el tipo de datos de un tensor al crearlo mediante el argumento dtype, como en a = torch.ones((2, 3), dtype=torch.int16), donde dtype se establece como un entero de 16 bits. También puedes cambiar el tipo de datos utilizando el método .to(), como en c = b.to(torch.int32), que convierte el tensor b a enteros de 32 bits. Los tipos de datos disponibles incluyen booleanos, enteros con y sin signo de 8, 16, 32 y 64 bits, y números de punto flotante de precisión simple, media, doble y bfloat. Usar tuplas para especificar la forma y agrupar argumentos hace que el código sea más legible.

````python
a = torch.ones((2, 3), dtype=torch.int16)          # tensor 2x3 de unos como enteros de 16 bits
print(a)

b = torch.rand((2, 3), dtype=torch.float64) * 20.0 # tensor 2x3 de floats aleatorios en [0,20)
print(b)

c = b.to(torch.int32)                              # convierte b a enteros de 32 bits (trunca decimales)
print(c)
````

**El broadcasting** en PyTorch es una técnica que permite realizar operaciones entre tensores de diferentes formas de manera automática y eficiente, sin necesidad de que tengan las mismas dimensiones. PyTorch extiende automáticamente las dimensiones más pequeñas para que coincidan con las dimensiones más grandes.

En este ejemplo, x es una matriz 1x3 y y es una matriz 3x1. El broadcasting se aplica automáticamente, y la operación de suma se realiza como si ambas matrices fueran del mismo tamaño (3x3), produciendo el resultado correcto.

````python
import torch

x = torch.tensor([[1, 2, 3]])
y = torch.tensor([[10], [20], [30]])
result = x + y  # Broadcasting: la fila se expande para coincidir con las columnas
print(result)
````

###Cambiar el número de dimensiones


Un caso en el que podría necesitar cambiar la cantidad de dimensiones es pasar una sola instancia como entrada a su modelo. Los modelos de PyTorch generalmente esperan *lotes* de entrada.

Por ejemplo, imagine tener un modelo que funcione con imágenes de 3 x 226 x 226, un cuadrado de 226 píxeles con 3 canales de color. Cuando lo cargues y lo transformes, obtendrás un tensor de forma ``(3, 226, 226)``. Sin embargo, su modelo espera una entrada de forma ``(N, 3, 226, 226)``, donde ``N`` es el número de imágenes en el lote. Entonces, ¿cómo se hace un lote de uno?

El método unsqueeze() agrega una dimensión de extensión 1. unsqueeze(0) lo agrega como una nueva dimensión cero - ¡ahora tienes un lote de uno!

````python
# Supongamos que tienes un tensor de forma (3, 226, 226)
single_instance = torch.randn((3, 226, 226))

# Añade una dimensión adicional para simular un lote de un elemento
batched_instance = single_instance.unsqueeze(0)

print("Tensor original:", single_instance.shape)
print("Tensor con una dimensión extra:", batched_instance.shape)
````

**squeeze()**

elimina dimensiones unitarias del tensor original, resultando en un tensor con una forma más compacta. Esto es útil para simplificar la estructura del tensor y suele ser utilizado para preparar datos antes de realizar operaciones o alimentarlos a modelos.

````python
import torch

# Supongamos que tienes un tensor de forma (1, 3, 1, 5)
original_tensor = torch.randn((1, 3, 1, 5))

# Utiliza el método squeeze para eliminar dimensiones unitarias
squeezed_tensor = original_tensor.squeeze()

print("Tensor original:", original_tensor.shape)
print("Tensor después de aplicar squeeze:", squeezed_tensor.shape)
````

**Reshape**  para cambiar la forma de un tensor tridimensional (output3d) a un tensor unidimensional (input1d) usamos esta función. La forma resultante es (6 * 20 * 20,). La última línea demuestra cómo realizar la misma operación utilizando torch.reshape como método en el módulo torch.







````python
import torch

# Supongamos que tienes un tensor de forma (6, 20, 20)
output3d = torch.rand(6, 20, 20)
print("Forma del tensor original:", output3d.shape)

# Utiliza el método reshape para cambiar la forma del tensor a 1D
input1d = output3d.reshape(6 * 20 * 20)
print("Forma después de reshape:", input1d.shape)

# También se puede llamar como método en el módulo torch
reshaped_tensor = torch.reshape(output3d, (6 * 20 * 20,))
print("Forma después de torch.reshape:", reshaped_tensor.shape)
````

DataLoaders
===================

Objetivo del código: Mostrar cómo se organiza el flujo de datos con PyTorch (Dataset + DataLoader) usando un conjunto de datos real, en este caso, el de diamantes. Esto es útil para entrenar modelos en lotes, barajar (shuffle) la data, etc.

````python
import torch
from torch.utils.data import TensorDataset, DataLoader
import pandas as pd
import seaborn as sns

# -----------------------------------------------------
# 1. Cargar el dataset de diamantes con Seaborn
# -----------------------------------------------------
df = sns.load_dataset("diamonds")

# -----------------------------------------------------
# 2. Separar la data en características (features) y objetivo (target)
#    - Elegimos columnas numéricas para features
#    - Usamos la columna 'price' como objetivo
# -----------------------------------------------------
features = df[['carat', 'depth', 'table', 'x', 'y', 'z']].dropna()
targets = df['price'].dropna()

# -----------------------------------------------------
# 3. Convertir a tensores de PyTorch
# -----------------------------------------------------
X_tensor = torch.tensor(features.values, dtype=torch.float32)
y_tensor = torch.tensor(targets.values, dtype=torch.float32)

# -----------------------------------------------------
# 4. Crear un TensorDataset a partir de los tensores
# -----------------------------------------------------
# TensorDataset asocia cada fila de X_tensor con la correspondiente fila de y_tensor.
dataset = TensorDataset(X_tensor, y_tensor)

# -----------------------------------------------------
# 5. Crear un DataLoader a partir del dataset
# -----------------------------------------------------
# El DataLoader se encarga de:
#   - Dividir la data en lotes (batch_size).
#   - Mezclar (shuffle=True) los datos para entrenamiento.
#   - Entregar la data de forma iterativa.
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

# -----------------------------------------------------
# 6. Usar el DataLoader en un bucle
# -----------------------------------------------------
# Cada vez que iteramos sobre "dataloader", recibimos un batch de X e y.
for batch_idx, (X_batch, y_batch) in enumerate(dataloader):
    print(f"Batch {batch_idx + 1}:") # se usa solo para mostrar el número de batch de forma más intuitiva para el usuario, no afecta el funcionamiento del código.
    print(f"  - X_batch.shape: {X_batch.shape}")  # Ejemplo: [32, 6]
    print(f"  - y_batch.shape: {y_batch.shape}")  # Ejemplo: [32]

    # Opcional: hacer algo con X_batch, y_batch (p.ej. pasar a un modelo)

    # Para el ejemplo, solo mostramos el primer batch
    # break
````

## Autograd # investigar!!! de que trata autograd -


En PyTorch, Autograd es una herramienta de diferenciación automática que rastrea y calcula automáticamente gradientes durante la ejecución de operaciones en tensores. Se activa por defecto, permitiendo la propagación automática de gradientes para realizar optimización de parámetros en modelos de aprendizaje profundo. Sin embargo, en casos donde no se requiere el cálculo de gradientes, se puede desactivar para mejorar la eficiencia computacional.

````python
import torch

# Crear un tensor con seguimiento automático de gradientes
x = torch.tensor([2.0], requires_grad=True)

# Realizar operaciones y calcular gradientes automáticamente
y = x**2
y.backward()  # Calcular el gradiente de y con respecto a x
print(x.grad)  # Imprimir el gradiente
````

En este código, se crea un tensor x con seguimiento automático de gradientes (requires_grad=True). Luego, se realiza la operación y = x**2 y se utiliza backward() para calcular automáticamente el gradiente de y con respecto a x. La derivada de y con respecto a x es 2 * x, y al evaluarlo en x=2.0, se obtiene 4.0. Por lo tanto, el resultado impreso tensor([4.]) representa el gradiente de y con respecto a x en el punto específico donde x es igual a 2.0.

````python
import torch #TAREA

# Crear un tensor sin seguimiento automático de gradientes
x = torch.tensor([2.0], requires_grad=False) # Mirar el error cuando no esta el gradiente  /TAREA

# Intentar realizar operaciones que requieren gradiente dará error
# y = x**2  # Esto generaría un error ya que x no está rastreando gradientes

# Se puede desactivar Autograd globalmente para mejorar la eficiencia
with torch.no_grad():
    y = x**2  # Ahora se permite realizar operaciones sin rastrear gradientes
````

````python
import torch

# Crear un tensor con seguimiento automático de gradientes
x = torch.tensor([2.0], requires_grad=True)

# Usar detach para crear un nuevo tensor sin seguimiento de gradientes
y = x.detach()
y
````
