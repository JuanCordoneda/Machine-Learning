```python
# Función de ganancia económica personalizada
def profit_score(y_true, y_pred, Ganancia_hit=5000, Costo_fp=2000, Costo_fn=1000, show_plots=True):
    """
    Calcula la ganancia total simulada y, si se desea, muestra:
    - Matriz de confusión (TP, FP, FN, TN)

    Retorna:
    --------
    profit : Ganancia total del modelo (USD)
    """

    # === Cálculo base ===
    tp = np.sum((y_true == 1) & (y_pred == 1))
    fp = np.sum((y_true == 0) & (y_pred == 1))
    fn = np.sum((y_true == 1) & (y_pred == 0))
    tn = np.sum((y_true == 0) & (y_pred == 0))
    profit = tp * Ganancia_hit - fp * Costo_fp - fn * Costo_fn

    if show_plots:
        # === Matriz de confusión ===
        fig_conf = go.Figure(data=go.Heatmap(
            z=[[tn, fp],
               [fn, tp]],
            x=["Predice No Éxito", "Predice Éxito"],
            y=["Real No Éxito", "Real Éxito"],
            colorscale="Blues",
            text=[[f"TN<br>{tn}", f"FP<br>{fp}"],
                  [f"FN<br>{fn}", f"TP<br>{tp}"]],
            texttemplate="%{text}",
            showscale=False
        ))
        fig_conf.update_layout(
            title="Matriz de Confusión (TP / FP / FN / TN)",
            title_font=dict(color="white", size=18),
            xaxis_title="Predicción del modelo",
            yaxis_title="Valor real",
            paper_bgcolor="#000000",   # Fondo del lienzo negro
            plot_bgcolor="#000000",    # Fondo del gráfico negro
            font=dict(color="white"),  # 🔹 Texto blanco en todo el gráfico
            xaxis=dict(color="white", showgrid=False),
            yaxis=dict(color="white", showgrid=False)
        )

        fig_conf.show()

    return profit


# === Nueva función de scoring con búsqueda de mejor umbral dinámico ===
def profit_scorer_dynamic(estimator, X, y_true):
    # Obtiene las probabilidades de predicción (columna de clase positiva)
    y_pred_proba = estimator.predict_proba(X)[:, 1]

    # Inicializa la mejor ganancia como menos infinito (para comparar)
    best_profit = -np.inf

    # Recorre posibles umbrales de decisión (de 0.01 a 0.49)
    for t in np.arange(0.01, 0.5, 0.01):
        # Convierte las probabilidades en predicciones binarias según el umbral
        y_pred = (y_pred_proba >= t).astype(int)

        # Calcula la ganancia económica para ese umbral
        profit = profit_score(y_true, y_pred, show_plots=False)

        # Si la ganancia mejora, actualiza el mejor valor encontrado
        if profit > best_profit:
            best_profit = profit

    # Devuelve la mejor ganancia obtenida entre todos los umbrales
    return best_profit
```

```python
# === PASO 1: CARGA Y LIMPIEZA DEL DATASET ===

# Cargar dataset proporcionado por un cliente del sector musical
# (por razones de confidencialidad, no se puede revelar el nombre ni la fuente exacta)
df = pd.read_csv("EDMHits.csv")

# Renombrar columnas a español para facilitar la interpretación
df.columns = [
    "Cancion", "Artista", "Album", "Año", "Duracion_ms", "Compas",
    "Bailabilidad", "Energia", "Tonalidad", "Volumen", "Modo", "Habla",
    "Acustica", "Instrumentalidad", "Presencia_en_vivo", "Valencia",
    "Tempo", "Popularidad"
]

df["Energia_Valencia"] = df["Energia"] * df["Valencia"]
df["Dance_Volumen"] = df["Bailabilidad"] * abs(df["Volumen"])
df["Tempo_Normalizado"] = df["Tempo"] / df["Duracion_ms"]
df["Acusticidad_Invertida"] = 1 - df["Acustica"]

# # 🔹 Eliminar columnas irrelevantes o redundantes
df = df.drop([
    "Cancion",           # Texto no numérico
    "Artista",           # Texto no útil para el modelo
    "Album",             # No aporta valor predictivo
    "Instrumentalidad",  # Muy correlacionada con "Acustica"
    "Bailabilidad",      # Alta correlación con "Energia" y "Valencia"
    "Presencia_en_vivo"  # Variable poco significativa para música electrónica
], axis=1)



# df = df.drop(["Cancion", "Artista", "Album"], axis=1)

# Crear variable objetivo: Éxito (1 = canción popular, 0 = no exitosa)
df["Exito"] = (df["Popularidad"] >= 60).astype(int)

df.head()
```

```python
# === PASO 3: DIVISIÓN TRAIN Y TEST ===

# Separar variables
X = df.drop(["Exito", "Popularidad"], axis=1)
y = df["Exito"]

# Escalado
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# División entrenamiento / test
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42, stratify=y)
```

```python
# === PASO 4: BUSQUEDA DE HIPERPARÁMETROS RANDOM FOREST (OPTIMIZADO PARA GANANCIA) ===

param_grid = {
    "n_estimators": [800, 1000, 1300, 1500],
    "max_depth": [10, 12, 14, 16],
    "min_samples_split": [2, 4, 6],
    "min_samples_leaf": [1, 2],
    "max_features": ["sqrt", "log2"]
}

grid_search = GridSearchCV(
    estimator=RandomForestClassifier(
        random_state=42,
        class_weight={0: 1, 1: 4.5}  # 🔥 AUMENTA TP → MENOS FN → SUBE GANANCIA
    ),
    param_grid=param_grid,
    scoring=profit_scorer_dynamic,
    cv=3,
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train, y_train)
best_model = grid_search.best_estimator_

print("\nMejores parámetros encontrados:")
print(grid_search.best_params_)
print(best_model)
print(f"Mejor ganancia promedio (3 folds): {grid_search.best_score_:.2f}")

# === Evaluación del mejor modelo ===

y_pred_proba_best = best_model.predict_proba(X_test)[:, 1]

profits = {}

# 🔥 Rango donde realmente sube la GANANCIA en tu dataset
for t in np.arange(0.08, 0.30, 0.01):
    y_pred_t = (y_pred_proba_best >= t).astype(int)
    profits[t] = profit_score(y_test, y_pred_t, show_plots=False)

best_threshold = max(profits, key=profits.get)
best_profit = profits[best_threshold]

print(f"\n🟦 Mejor umbral en test: {best_threshold:.2f}")
print(f"💰 Ganancia máxima en test: {best_profit:.2f}")
```

```python
# === MODELO FINAL MAX-GANANCIA (CON LOS PARÁMETROS REALES ÓPTIMOS) ===

best_params_profit = {
    "n_estimators": 1300,
    "max_depth": 10,
    "max_features": "sqrt",
    "min_samples_leaf": 1,
    "min_samples_split": 2,
    "class_weight": {0: 1, 1: 4.5},
    "criterion": "gini",
    "random_state": 42
}

rf_profit = RandomForestClassifier(**best_params_profit)
rf_profit.fit(X_train, y_train)

# Probabilidades directas
y_pred_proba = rf_profit.predict_proba(X_test)[:, 1]

# 🔥 Usamos el umbral óptimo
best_threshold = 0.22
y_pred_final = (y_pred_proba >= best_threshold).astype(int)

# Evaluación económica
ganancia = profit_score(y_test, y_pred_final)
print(f"💰 GANANCIA FINAL DEL MODELO (CORRECTA): {ganancia:.2f}")

from sklearn.metrics import confusion_matrix, precision_score, recall_score

cm = confusion_matrix(y_test, y_pred_final)
print("Matriz de confusión:\n", cm)
print(f"Precisión: {precision_score(y_test, y_pred_final):.3f}")
print(f"Sensibilidad (Recall): {recall_score(y_test, y_pred_final):.3f}")
```

🧠 ¿POR QUÉ ESTE ES EL PUNTO ÓPTIMO?
✔ Recall altísimo = capturás casi todos los éxitos

Esto dispara la ganancia porque los verdaderos positivos valen muchísimo.

✔ Precisión intermedia = te permite contener los falsos positivos

Si subís el threshold → baja recall → se desploma la ganancia.
Si bajás el threshold → suben FP → se desploma la ganancia.

0.22 es el único punto donde ambos balancean.

✔ Random Forest con min_samples_leaf=1 y class_weight=4.5

Genera árboles más finos y sensibles a la clase minoritaria → sube recall sin perder demasiado en precisión.

✔ Umbral bajo pero NO tan bajo

0.14 daba mucha ganancia pero destrozaba precision.
0.30 daba buena precision pero recall ridículo.
0.22 es el “saddle point”.

⭐ La precisión no va a subir más SOLO con tuning

Para mejorarla, necesitás nuevos features, no hiperparámetros.

Ya lo vimos:

XGBoost: precisión 0.35

RF optimizado: precisión 0.35

RF sin leaf=1: precisión 0.32

Modelos calibrados: precisión 0.19–0.25

Threshold >0.3: recall se hunde

Todo converge al mismo techo: 0.35 ± 0.05.

Ese es el límite natural con los features actuales.

```python
# === PASO 7: XGBOOST AUTO-TUNEADO PERO USANDO TU LÓGICA DE UMBRAL ===

from xgboost import XGBClassifier
from sklearn.model_selection import RandomizedSearchCV
import numpy as np
from sklearn.metrics import confusion_matrix, precision_score, recall_score

# 1) Espacio de búsqueda
param_dist = {
    "n_estimators": [200, 300, 400, 600, 800],
    "max_depth": [3, 4, 5, 6],
    "learning_rate": [0.05, 0.1, 0.2],
    "subsample": [0.7, 0.8, 0.9],
    "colsample_bytree": [0.6, 0.7, 0.8, 0.9],
    "gamma": [0, 0.1, 0.2],
    "min_child_weight": [1, 2, 3, 4],
}

# 2) Optimización logloss
xgb = XGBClassifier(
    random_state=42,
    eval_metric="logloss"
)

random_search = RandomizedSearchCV(
    estimator=xgb,
    param_distributions=param_dist,
    n_iter=40,
    scoring="neg_log_loss",
    cv=3,
    verbose=1,
    n_jobs=-1,
    random_state=42
)

random_search.fit(X_train, y_train)
best_xgb = random_search.best_estimator_

print("\nMejores hiperparámetros (LogLoss):")
print(random_search.best_params_)

# 3) Probabilidades del modelo
y_proba = best_xgb.predict_proba(X_test)[:, 1]

# 4) Buscar mejor umbral (GANANCIA real)
profits = {}
for t in np.arange(0.01, 0.50, 0.01):
    preds = (y_proba >= t).astype(int)
    profits[t] = profit_score(y_test, preds, show_plots=False)

best_t = max(profits, key=profits.get)
best_profit = profits[best_t]

print(f"\n🔥 Mejor umbral: {best_t:.2f}")
print(f"💰 Ganancia con XGB auto-tuneado + umbral: {best_profit:.2f}")

# 5) Evaluación final
y_pred_final = (y_proba >= best_t).astype(int)

cm = confusion_matrix(y_test, y_pred_final)
prec = precision_score(y_test, y_pred_final)
rec = recall_score(y_test, y_pred_final)

print("\nMatriz de confusión:")
print(cm)
print(f"Precisión: {prec:.3f}")
print(f"Recall: {rec:.3f}")
```
